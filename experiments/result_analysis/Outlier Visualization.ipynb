```python
import os

import folium
import io
from PIL import Image
```

```python
# Build the default map for a specific location
def create_map(geojson, location, zoom_start=14):
    map = folium.Map(location=location, zoom_start=zoom_start, png_enabled=True, zoomControl=False, tiles='CartoDB Voyager')

    color_list = ["orange", "red", "blue"]
    icon_list = ["asterisk", "unchecked", "chevron-down"]

    for i, feature in enumerate(geojson['features']):
        # get name of approach

        # add lines
        trail_coordinates = []
        for f in feature['geometry']['coordinates']:
            trail_coordinates.append((f[1],f[0]))

        folium.PolyLine(trail_coordinates, color=color_list[i], weight=5,).add_to(map)

    return map
```
