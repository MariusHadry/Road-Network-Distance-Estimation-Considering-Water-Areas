```python
# Build the default map for a specific location
def create_map(geojson, location, zoom_start=14, line_weight=5):
    map = folium.Map(location=location, zoom_start=zoom_start, png_enabled=True, zoomControl=False,
                     tiles='https://{s}.basemaps.cartocdn.com/light_nolabels/{z}/{x}/{y}{r}.png',
                     attr='&copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> contributors &copy; <a href="https://carto.com/attributions">CARTO</a>'
                     # tiles='CartoDB Voyager'
                     )

    color_list = ["orange", "red", "blue", "black"]
    icon_list = ["asterisk", "unchecked", "chevron-down"]

    for i, feature in enumerate(geojson['features']):
        # get name of approach

        # add lines
        trail_coordinates = []
        for f in feature['geometry']['coordinates']:
            trail_coordinates.append((f[1],f[0]))

        folium.PolyLine(trail_coordinates, color=color_list[i], weight=line_weight,).add_to(map)

    return map
```
